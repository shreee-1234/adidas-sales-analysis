```python
import pandas as pandas
import numpy as np 
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_excel("adidas_sales.csv", header=4)
df = df.drop(columns=['Unnamed: 0'])
print(df.head())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
print("Total Sales: $", round(df['Total Sales'].sum(), 2))
```

```python
print("Total Profit: $", round(df['Operating Profit'].sum(), 2))
```

```python
top_retailers = (
    df.groupby('Retailer')['Total Sales']
      .sum()
      .sort_values(ascending=False)
)

print(top_retailers)
```

```python
import matplotlib.pyplot as plt

top_retailers.plot(kind='bar', figsize=(10,5))
plt.title('Total Sales by Retailer')
plt.ylabel('Sales')
plt.xlabel('Retailer')
plt.xticks(rotation=45)
plt.show()
```

```python
sales_region = (
    df.groupby('Region')['Total Sales']
      .sum()
      .sort_values(ascending=False)
)

print(sales_region)
```

```python
sales_region.plot(kind='bar', figsize=(8,5))
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```

```python
top_products = (
    df.groupby('Product')['Total Sales']
      .sum()
      .sort_values(ascending=False)
)

print(top_products)
```

```python
top_products.plot(kind='bar', figsize=(10,5))
plt.title('Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
sales_region = (
    df.groupby('Region')['Total Sales']
      .sum()
)

plt.figure(figsize=(8,8))
plt.pie(
    sales_region,
    labels=sales_region.index,
    autopct='%1.1f%%'
)
plt.title('Sales Distribution by Region')
plt.show()
```

```python
sales_method = (
    df.groupby('Sales Method')['Total Sales']
      .sum()
)

plt.figure(figsize=(8,8))
plt.pie(
    sales_method,
    labels=sales_method.index,
    autopct='%1.1f%%'
)
plt.title('Sales Distribution by Sales Method')
plt.show()
```

```python
top5_retailers = (
    df.groupby('Retailer')['Total Sales']
      .sum()
      .sort_values(ascending=False)
      .head(5)
)

plt.figure(figsize=(8,8))
plt.pie(
    top5_retailers,
    labels=top5_retailers.index,
    autopct='%1.1f%%'
)
plt.title('Top 5 Retailers Sales Share')
plt.show()
```
